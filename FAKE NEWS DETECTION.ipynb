```python
import pandas as pd
import matplotlib.pyplot as plt

# Load datasets
fake = pd.read_csv("Fake.csv")
true = pd.read_csv("True.csv")

# Add labels
fake["label"] = "Fake"
true["label"] = "Real"

# Combine datasets
data = pd.concat([fake, true])

# Count fake and real news
counts = data["label"].value_counts()

# Calculate percentages
percentages = (counts / len(data)) * 100

# Print percentages
print("Fake News Percentage:")
print(percentages["Fake"])

print("\nReal News Percentage:")
print(percentages["Real"])

# Plot graph
counts.plot(kind='pie', autopct='%1.1f%%')

plt.title("Fake vs Real News Distribution")
plt.ylabel("")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
fake = pd.read_csv("Fake.csv")
true = pd.read_csv("True.csv")
fake["label"] = "Fake"
true["label"] = "Real"
data = pd.concat([fake, true])
counts = data["label"].value_counts()
percentages = (counts / len(data)) * 100
print("========== NEWS ANALYSIS REPORT ==========\n")
print("Total News Articles:", len(data))
print("\nFake News Count:", counts["Fake"])
print("Real News Count:", counts["Real"])
print("\nFake News Percentage: {:.2f}%".format(percentages["Fake"]))
print("Real News Percentage: {:.2f}%".format(percentages["Real"]))
print("\n========== ANALYSIS ==========")
if percentages["Fake"] > percentages["Real"]:
    print("The dataset contains more fake news articles than real news articles.")
else:
    print("The dataset contains more real news articles than fake news articles.")

print("This analysis helps understand the distribution of misinformation in digital media.")


plt.figure(figsize=(6,5))
counts.plot(kind='bar')

plt.title("Fake vs Real News Count")
plt.xlabel("News Type")
plt.ylabel("Number of Articles")

plt.show()


plt.figure(figsize=(6,6))
counts.plot(kind='pie', autopct='%1.1f%%')

plt.title("Fake vs Real News Distribution")
plt.ylabel("")

plt.show()
```
